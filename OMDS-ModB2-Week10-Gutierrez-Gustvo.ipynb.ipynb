### How does random forest work?

A random forest is a set of decision trees.  Let's say we use n_estimators = 100 decision trees, and we have ten features.  Then each decision tree is built so that a given branch gets about three (sqrt(10)) features to choose from.  It starts from the top of the tree, selects the best feature, then proceeds to the next nodes.  The best feature is the one that minimizes the mean squared error, which must be computed for both branches, weighted according to the probability of each branch, and then summed over both branches.

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import make_scorer
```

Here, we build three series of random numbers between 0 and 1.  We use a somewhat arbitrary formula to construct the synthetic data.  The fourth value (the target) equals the first series minus twice the second, with the whole then being multiplied by the third series.  If we stopped here, the target would be completely dependent on the three feature series - it would be easy to predict.  However, we then add a random normal distribution to make the prediction harder.

There are a train series and a test series, with different features ("a" and "b"), and with different random normal data added.  This tests whether our model can predict new data.  We fit to the "a" features using the train_target, then predict the test_target using the "b" features.

```python
np.random.seed(0)
num_points = 10000
feature_1a = np.random.random(size = num_points) * 3
feature_2a = np.random.random(size = num_points) * 3
feature_3a = np.random.random(size = num_points) * 3
train_target = (feature_1a - 2 * feature_2a) * feature_3a + np.random.normal(size = num_points)
feature_1b = np.random.random(size = num_points) * 3
feature_2b = np.random.random(size = num_points) * 3
feature_3b = np.random.random(size = num_points) * 3
test_target = (feature_1b - 2 * feature_2b) * feature_3b + np.random.normal(size = num_points)
train_df = pd.DataFrame({"f1": feature_1a, "f2": feature_2a, "f3": feature_3a})
test_df = pd.DataFrame({"f1": feature_1b, "f2": feature_2b, "f3": feature_3b})
rf = RandomForestRegressor()
rf.fit(train_df.values, train_target)
```

As expected, the RMSE is smallest when predicting the training data, larger when predicting the testing data, and largest when comparing random (shuffled) data instead of using a prediction.  In the last case, there is no attempt to make a good prediction.  The small RMSE when predicting the training data indicates that the model has "memorized" the training data to some degree.

```python
print("RMSE when predicting training data:")
print(root_mean_squared_error(rf.predict(train_df.values), train_target))
print("RMSE when predicting testing data:")
print(root_mean_squared_error(rf.predict(test_df.values), test_target))
test_target_shuffle = test_target.copy()
np.random.shuffle(test_target_shuffle)
print("RMSE with random predictions:")
print(root_mean_squared_error(test_target_shuffle, test_target))
```

Next, we perform a Grid Search over various parameters.  We've used a small number of parameters here, but you could add more.  The first parameter, n_estimators, shows the number of decision trees in the random forest.  The second, max_depth, is the maximum depth of each tree.  By picking three n_estimators values (10, 33, 100) and three max_depth values (2, 4, None), we allow for nine parameter combinations.  For example, (n_estimators = 10, max_depth = 2) is a combination.  The grid search tries all nine combinations and uses cross validation to choose the best one.  This is done below.

### What is cross validation?

Cross validation is a way to find out which parameter combination is the best.  With 5-fold cross validation (the default) we divide the data into five equal-sized groups.  Four of them are used as training data and the fifth is testing data.  The score (in this case the root mean squared error) on the testing data is recorded.  This is done for each split: that is, each fifth is taken out and used as testing data; then, the five root mean squared errors (from taking out the five fifths, one at a time) are averaged.

We now have root mean squared error value for each parameter combination, allowing us to choose the best combination.

```python
np.random.seed(0)
gcv = GridSearchCV(estimator = RandomForestRegressor(), param_grid = {"n_estimators": [10, 33, 100], "max_depth": [2, 4, None]}, \
                   scoring = "neg_root_mean_squared_error")
gcv.fit(train_df.values, train_target)
```

```python
print(gcv.best_estimator_)
print(gcv.best_params_)
print(gcv.best_score_)
```

```python
gcv.cv_results_["mean_test_score"] # There are nine scores, one for each of the 3 x 3 hyperparameter combinations
```

### Why does n_estimators grow like 10, 33, 100 instead of like 1, 50, 100?

First of all, we can make it grow however we want.  The question is why we chose 10, 33, 100.  The idea here is that when you have no idea what is the appropriate number, you should pick guesses that grow exponentially.  For example, suppose you are trying to guess the height of a particular tree.  You look online and find that redwoods grow to 400 feet high, while bonsai trees may be as short as six inches.  If you get three guesses, you could guess: one foot, 200 feet, 400 feet.  These guesses would be linearly spaced out.  However, the 200 foot guess feels wrong - it's still very tall.  Most trees don't grow that tall.  Instead, let's make a geometric progression:

```python
list(map(round, [400**0, 400**0.25, 400**0.5, 400**0.75, 400*1]))
```

Now the guesses are four feet, 20 feet, and 89 feet.  (We'll drop the initial 1 and the final 400.) That feels more plausible: the middle value, 20 feet, is representative of many actual trees in a way that the middle 200 foot guess wasn't.

# 3. Homework

In this exercise, you will try to predict the change of one stock based on the change in ten other stocks.

```python
def make_X_y(alpha = 0.5):
    df_hwk = pd.DataFrame()
    num_samples_hwk = 10000
    common_odd = np.random.normal(size = num_samples_hwk)
    common_even = np.random.normal(size = num_samples_hwk)
    common_three = np.random.normal(size = num_samples_hwk)
    for n in range(11):
        common = np.zeros(num_samples_hwk)
        if n % 2 == 0:
            common += common_even
        if n % 2 == 1:
            common += common_odd
        if n % 3 == 0:
            common += common_three
            common /= 2
        df_hwk["Stock_" + str(n)] = np.random.normal(size = num_samples_hwk) + alpha * common
    X = df_hwk.values[:,:-1]
    y = df_hwk.values[:,-1]
    return X, y
```

The above code creates an X with ten columns, corresponding to ten different stocks.  The y is the eleventh stock.  The stocks have some commonalities: the even numbered stocks are related, as are the odds.  The stocks that are multiples of three are all related, too.  Random forest should be able to sort through all this and decide that the eleventh stock is related to the odd numbered, non-divisible-by-three stocks, and predict accordingly.  The prediction should presumably be, simply, to take columns 1, 5, and 7 and average them, as they are odd-numbered and not divisible by three; so perhaps this task would be easier to do with a linear regression, but we'll use a random forest anyway.  The value "alpha" determines how important the common element is.  If alpha = 0, each stock is its own column of normally-distributed numbers with no relation to the others.  If alpha = 10, then the great majority of the variation in the stock values comes from the common elements.

### My Prediction: alpha = 0 vs alpha = 10

I predict that **alpha = 10 will result in much better predictions** than alpha = 0.

**Why?** The `alpha` parameter controls how much of each stock's value comes from the shared common factors versus pure random noise. When alpha = 0, every stock is just independent random noise — there is no relationship between any of the stocks, so there is nothing for the model to learn. The ten feature stocks have zero predictive power over the target stock. When alpha = 10, the common factors (odd/even, divisible-by-three) dominate the stock values, creating strong correlations that the random forest can exploit to make accurate predictions.

In short: alpha = 0 means "no signal, all noise" and alpha = 10 means "strong signal, little noise." A model can only predict well when there is an actual pattern to find.

```python
from scipy.stats import pearsonr

# --- Alpha = 0 (no signal) ---
np.random.seed(42)
X_train_0, y_train_0 = make_X_y(alpha=0)
X_test_0, y_test_0 = make_X_y(alpha=0)

rf_0 = RandomForestRegressor(n_estimators=100, random_state=42)
rf_0.fit(X_train_0, y_train_0)
pred_0 = rf_0.predict(X_test_0)
corr_0, _ = pearsonr(pred_0, y_test_0)
rmse_0 = root_mean_squared_error(pred_0, y_test_0)

# --- Alpha = 10 (strong signal) ---
np.random.seed(42)
X_train_10, y_train_10 = make_X_y(alpha=10)
X_test_10, y_test_10 = make_X_y(alpha=10)

rf_10 = RandomForestRegressor(n_estimators=100, random_state=42)
rf_10.fit(X_train_10, y_train_10)
pred_10 = rf_10.predict(X_test_10)
corr_10, _ = pearsonr(pred_10, y_test_10)
rmse_10 = root_mean_squared_error(pred_10, y_test_10)

print("=== Alpha = 0 (no signal) ===")
print(f"  Correlation: {corr_0:.4f}")
print(f"  RMSE:        {rmse_0:.4f}")
print()
print("=== Alpha = 10 (strong signal) ===")
print(f"  Correlation: {corr_10:.4f}")
print(f"  RMSE:        {rmse_10:.4f}")
```

### Results Commentary

As predicted, **alpha = 10 produces a much higher correlation** between predicted and actual values, confirming that when the common signal between stocks is strong, the random forest can learn the relationships and make accurate predictions. With alpha = 0, the correlation is near zero — the model is essentially guessing, because there is no real pattern linking the feature stocks to the target stock.

The RMSE values are harder to compare directly between the two alphas because the magnitude of the data itself changes with alpha (alpha = 10 data has much larger values). This is exactly why the assignment recommended using correlation instead of RMSE for comparing across different alpha values — correlation is scale-independent.

### RandomizedSearchCV

Instead of GridSearchCV (which tries every combination), RandomizedSearchCV samples a fixed number of random parameter combinations from specified distributions. This is more efficient when the hyperparameter space is large, because you get a good search with fewer model fits.

```python
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

np.random.seed(42)

# Define parameter distributions to sample from
param_distributions = {
    "n_estimators": [10, 33, 100, 200],
    "max_depth": [2, 4, 8, None],
    "min_samples_split": randint(2, 20),
    "min_samples_leaf": randint(1, 10)
}

# Use the alpha=10 data for tuning (where there is actual signal)
rcv = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_distributions=param_distributions,
    n_iter=20,  # try 20 random combinations
    scoring="neg_root_mean_squared_error",
    cv=5,
    random_state=42
)
rcv.fit(X_train_10, y_train_10)

print("Best parameters found:")
print(rcv.best_params_)
print(f"\nBest CV score (neg RMSE): {rcv.best_score_:.4f}")
print(f"\nTest set RMSE: {root_mean_squared_error(rcv.predict(X_test_10), y_test_10):.4f}")
print(f"Test set correlation: {pearsonr(rcv.predict(X_test_10), y_test_10)[0]:.4f}")
```

### Scoring Parameter Research: R-squared (r2)

I chose to explore **R-squared (r2)** as an alternative scoring metric.

**What is R-squared?** R-squared measures the proportion of variance in the target variable that is explained by the model's predictions. It ranges from negative infinity to 1.0:
- **R2 = 1.0** means the model perfectly explains all variation in the target
- **R2 = 0.0** means the model is no better than simply predicting the mean of the target every time
- **R2 < 0** means the model is *worse* than just predicting the mean (a bad model)

**Why is it useful?** Unlike RMSE, R-squared gives you an intuitive sense of "what fraction of the pattern did the model capture?" An R2 of 0.85 means the model explains 85% of the variance. This makes it easy to communicate model quality to non-technical stakeholders. However, R2 can be misleading if the data has outliers, and it always increases (or stays the same) as you add more features — which can encourage overfitting if you are not careful.

```python
from sklearn.metrics import r2_score

# Run RandomizedSearchCV again using R2 scoring
rcv_r2 = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=42),
    param_distributions=param_distributions,
    n_iter=20,
    scoring="r2",
    cv=5,
    random_state=42
)
rcv_r2.fit(X_train_10, y_train_10)

print("=== RandomizedSearchCV with R2 scoring ===")
print(f"Best parameters: {rcv_r2.best_params_}")
print(f"Best CV R2 score: {rcv_r2.best_score_:.4f}")

# Compare R2 on test set for both alpha values
pred_10_tuned = rcv_r2.predict(X_test_10)
pred_0_default = rf_0.predict(X_test_0)

print(f"\nR2 on test set (alpha=10, tuned): {r2_score(y_test_10, pred_10_tuned):.4f}")
print(f"R2 on test set (alpha=0):          {r2_score(y_test_0, pred_0):.4f}")
print("\nAs expected, R2 is high for alpha=10 (strong signal) and near 0 or negative for alpha=0 (no signal).")
```

# 5. Full Explanation of Decisions and Reasoning

## Lesson Recap

This week's lesson introduced **Random Forest Regression**, a model that builds many decision trees and averages their predictions. Key concepts covered:

- **Random Forest**: Each tree sees a random subset of features at each split (sqrt of total features by default). This diversity among trees reduces overfitting compared to a single decision tree.
- **RMSE (Root Mean Squared Error)**: Measures prediction error in the same units as the target. Lower is better. The lesson showed that RMSE is lowest on training data (the model "memorizes" it), higher on test data, and highest with random guesses.
- **Cross-Validation (5-fold)**: Splits training data into 5 parts, trains on 4, tests on the 5th, and rotates. This gives a reliable estimate of how well the model generalizes without needing a separate validation set.
- **GridSearchCV**: Tries every combination of hyperparameters. Good for small search spaces but expensive for large ones.
- **Geometric spacing**: When unsure of the right scale for a hyperparameter, use exponentially-spaced values (10, 33, 100) instead of linear ones (1, 50, 100), because this covers a wider range more efficiently.

## Weekly Graph Question

The RMSE-vs-training-size plot demonstrates **learning curves**: more training data leads to better test performance, but with diminishing returns. This is fundamental to understanding when collecting more data will help versus when you need a better model.

## Homework Reasoning

### Alpha Comparison
- **Why alpha=10 beats alpha=0**: Alpha controls signal-to-noise ratio. Stock 10 (the target, odd and not divisible by 3) shares the `common_odd` factor with stocks 1, 3, 5, 7, 9. At alpha=0, there is zero shared signal, so prediction is impossible. At alpha=10, the shared factors are 10x the noise, making prediction straightforward.
- **Why correlation over RMSE**: The assignment noted that RMSE depends on data magnitude. Alpha=10 data has larger values, so RMSE is naturally larger even with a good model. Correlation is scale-free and directly measures "how well do predictions track actuals."

### RandomizedSearchCV
- **Why RandomizedSearchCV instead of GridSearchCV**: With 4 hyperparameters and multiple values each, a full grid would require hundreds of fits. RandomizedSearchCV samples 20 random combinations (n_iter=20), which is computationally cheaper and often finds near-optimal parameters because it can explore a wider range of values.
- **Parameter choices**: I included `min_samples_split` and `min_samples_leaf` in addition to `n_estimators` and `max_depth` because they control tree complexity and help prevent overfitting.

### R-squared Scoring
- **Why R2**: It gives an intuitive "percentage of variance explained" interpretation. An R2 of 0.90 means the model captures 90% of the pattern. This is easier to communicate than RMSE, which requires knowing the scale of the data to interpret.
- **Key insight**: R2 can go negative (worse than predicting the mean), which is exactly what happens with alpha=0 — the model's attempts to find patterns in pure noise actually hurt its predictions.

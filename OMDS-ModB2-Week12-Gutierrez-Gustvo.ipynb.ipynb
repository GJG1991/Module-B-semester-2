# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

# Answer

**Side-by-side subplots (Plot 1):**
- *Advantage:* Each week gets its own space, so you can clearly see the shape of each week's pattern without one line hiding the other. It is easier to read individual daily values.
- *Disadvantage:* It is harder to directly compare the two weeks because the plots are separated. Your eyes have to jump back and forth. Also, with `wspace=0` the y-axis labels only appear on the left panel which can be confusing for the right panel.

**Overlaid on the same axes (Plot 2):**
- *Advantage:* Makes it very easy to compare the two weeks directly since the lines share the same axes. You can immediately see that Week 1 starts high and trends down while Week 2 starts low and trends up the patterns are roughly opposite.
- *Disadvantage:* If the lines overlap or cross frequently it can become cluttered and hard to read. With only two lines it works fine, but this approach does not scale well to many weeks of data.

Overall, the overlaid plot is better for this comparison because the goal is to see how the two weeks differ, and having shared axes makes that comparison immediate.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(2008, 400, s=80, color='#333333', zorder=5)
ax.scatter(2008, 350, s=80, color='#2F5597', zorder=5)

ax.text(2008.15, 400, 'A', fontsize=11, va='center', fontweight='bold', color='#333333')
ax.text(2008.15, 350, 'B', fontsize=11, va='center', fontweight='bold', color='#2F5597')

ax.set_xlim(2007.5, 2014.5)
ax.set_ylim(0, 500)
ax.set_xticks(range(2008, 2015))
ax.set_yticks([0, 100, 200, 300, 400, 500])
ax.set_yticklabels(['$0', '$100', '$200', '$300', '$400', '$500'])
ax.set_xlabel('Year', fontsize=10)
ax.set_ylabel('Average price', fontsize=10)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

fig.suptitle('Products A and B were launched in 2008 at price points of $360+',
             fontsize=12, fontweight='bold', x=0.12, ha='left', y=0.95)
ax.set_title('Retail price over time', fontsize=10, color='gray', loc='left', pad=10)

plt.tight_layout()
plt.subplots_adjust(top=0.85)
plt.show()
```
